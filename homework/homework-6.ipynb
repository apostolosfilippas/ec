#### **1. The informational content of public and private feedback**
In your online labor market, employers (users who post jobs) transact with workers (freelancers). For example, se [Upwork](http://upwork.com/), [Taskrabbit](http://taskrabbit.com/), [Freelancer](https://www.freelancer.com/), and so on.

Unfortunately, your public feedback scores suffer from severe reputation inflation (see HW1, Lecture 5, and practically any peer-to-peer marketplace).  To examine the consequences of reputation inflation, the platform introduced a new private feedback feature: in addition to giving public feedback, employers were also asked for “private” feedback. The private feedback was elicited by asking employers “Would you hire this worker again in the future?” after each transaction, to which the employers were asked to select one of four pre-populated answers. Importantly, the platform told the employers that this private feedback would be kept anonymous (the identity of the rating employer would not be shared with anyone) and hidden (the feedback would only be collected by the platform for internal evaluation purposes, and not shared with the other users, including future employers or the rated worker).

Because of this platform intervention, you now have access to both private and public feedback information for the same employer-worker transaction. Notice that public feedback, which is given as a numerical score, is also given in “bin” format. Each bin simply represents the range in which the score is contained. For example, all scores from 1 to 1.25 belong to the same bin [1,1.25].
