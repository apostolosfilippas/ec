```python
from edsl import ScenarioList, QuestionMultipleChoice

prices = ScenarioList.from_list("price", [16,20,40,100])
political_views = ScenarioList.from_list("political_view", [
        "socialist",
        "leftist",
        "liberal",
        "moderate",
        "libertarian",
        "conservative"
        ])

q = QuestionMultipleChoice(
    question_name="shovel_experiment",
    question_text="""
    A hardware store has been selling snow shovels for $15.
    The morning after a large snowstorm, the store raises the price to ${{price}}.
    You are a {{political_view}}.
    How would you rate this action?
    """,
    question_options=["Completely Fair", "Acceptable", "Unfair", "Very Unfair"],
)

scenarios = prices * political_views

responses = q.by(scenarios).run(n=100)
```

```python
responses.select("scenario.political_view", "scenario.price", "answer.*")
```

```python
df = responses.to_pandas()
df.columns
```

```python
# Statsmodels is installed when you run `uv sync` in the course folder.
```

```python
import statsmodels.formula.api as smf

mapping = {
    "Completely Fair": 4,
    "Acceptable": 3,
    "Unfair": 2,
    "Very Unfair": 1
}

df["fair_num"] = df["answer.shovel_experiment"].map(mapping)
df["price"] = df["scenario.price"]
df["political_view"] = df["scenario.political_view"]

model = smf.ols(
    formula="fair_num ~ price + C(political_view)",
    data=df
).fit()

print(model.summary())

```
